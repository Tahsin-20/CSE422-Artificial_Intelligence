````python
# -*- coding: utf-8 -*-
"""8queens_genetic_algorithm.ipynb

Automatically generated by Colab.

Original file is located at
    https://colab.research.google.com/drive/1qBPyKuE41CYlx3fRZHh_PmbTTZlj_Meu
"""

import random
N = 8
POP_SIZE = 8
MAX_GENERATIONS = 1000
MUTATION_RATE = 0.1
MAX_PAIRS = N * (N - 1) // 2

"""## 2. Make a random board

##Your task : Complete the **fitness** function
"""

def random_individual():
  popu=[]
  for i in range (N):
    popu.append(random.randint(1,N))
  return popu


def fitness(board):
    non_attacking = 0
    attacking=0
    for i in range(N):
        for j in range(i + 1, N):
            if board[i]==board[j]:
              attacking+=1
            if abs(board[i]-board[j])==abs(i-j):
              attacking+=1
        non_attacking=MAX_PAIRS-attacking
    return non_attacking

"""## 3. Selection — roulette wheel"""

def select_parent(population):
    scores = [fitness(b) for b in population]
    total = sum(scores)
    pick = random.uniform(0, total)
    running = 0
    for board, score in zip(population, scores):
        running += score
        if running >= pick:
            return board
    return population[-1]

"""## 4. Crossover and mutation"""

def crossover(parent1, parent2):
    point = random.randint(1, N - 1)
    child1 = parent1[:point] + parent2[point:]
    child2 = parent2[:point] + parent1[point:]
    return child1, child2

def mutate(board):
    column = random.randint(0, N - 1)
    board[column] = random.randint(1, N)
    return board

"""## 5. The Genetic Algorithm loop"""

def genetic_algorithm():
    population=[]
    for i in range (POP_SIZE):
      population.append(random_individual())
    best = max(population, key=fitness)

    for generation in range(MAX_GENERATIONS):
        population.sort(key=fitness, reverse=True)
        new_population = [population[0]]

        while len(new_population) < POP_SIZE:
            p1 = select_parent(population)
            p2 = select_parent(population)
            c1, c2 = crossover(p1, p2)
            if random.random() < MUTATION_RATE:
                c1 = mutate(c1)
            if random.random() < MUTATION_RATE:
                c2 = mutate(c2)
            new_population.append(c1)
            if len(new_population) < POP_SIZE:
                new_population.append(c2)

        population = new_population
        current_best = max(population, key=fitness)
        if fitness(current_best) > fitness(best):
            best = current_best

        if fitness(best) == MAX_PAIRS:
            print("Solution found in generation", generation)
            break

    return best

"""## 6. Run it"""

solution = genetic_algorithm()
print("Best board:", solution)
print("Non-attacking pairs:", fitness(solution), "out of", MAX_PAIRS)

"""## 7. See the board as a grid"""

def print_board(board):
    for row in range(N, 0, -1):
        line = ""
        for col in range(N):
            line += " Q" if board[col] == row else " ."
        print(line)
    print()

print_board(solution)

"""## 8. Your task : Write the **two-point crossover** function
```
Parent A : [a1 | a2 a3 a4 | a5 a6 a7 a8]
Parent B : [b1 | b2 b3 b4 | b5 b6 b7 b8]   cuts after position 1 and after position 4
Child 1  : [a1 | b2 b3 b4 | a5 a6 a7 a8]
Child 2  : [b1 | a2 a3 a4 | b5 b6 b7 b8]
```

**Rules:**
1. Pick `point1` randomly between `1` and `N-2`.
2. Pick `point2` randomly **after** `point1` (from `point1 + 1` up to `N-1`) — the second cut must always come after the first.
3. Build the two children by keeping the outer parts of one parent and the middle of the other.

Fill in the four lines marked `TODO`, then run the test cell.
"""

def crossover_two_point(parent1, parent2):
    point1 = random.randint(1,N-2)
    point2 = random.randint(point1+1,N-1)

    child1 = parent1[:point1]+parent2[point1:point2]+parent1[point2:]  # TODO
    child2 = parent2[:point1]+parent1[point1:point2]+parent2[point2:]  # TODO
    return child1, child2

"""**Test your function.**"""

a = [1, 1, 1, 1, 1, 1, 1, 1]
b = [8, 8, 8, 8, 8, 8, 8, 8]

c1, c2 = crossover_two_point(a, b)
print("Child 1:", c1)
print("Child 2:", c2)
````

````python
import random
import math
blocks=[("ALU",5,5),("Cache",7,4),("Control Unit",4,4),("Register Flie",6,6),("Decoder",5,3),("Floating Unit",5,5)]
connections=[(3,0),(2,0),(1,0),(3,5),(1,4),(4,5)]
Max_pop=6
def fitness(chromosome):
  overlap=overlapping(chromosome)
  wiring=wire_length(chromosome)
  bounding=area(chromosome)

  return -(1000*overlap+2*wiring+bounding)


def overlapping (chromosome):
  oc=0
  for i in range (len(blocks)):
    for j in range (i+1,len(blocks)):
      x1,y1=chromosome[i]
      _,w1,h1=blocks[i]
      left1=x1
      right1=left1+w1
      bottom1=y1
      top1=bottom1+h1

      x2,y2=chromosome[j]
      _,w2,h2=blocks[j]
      left2=x2
      right2=left2+w2
      bottom2=y2
      top2=bottom2+h2

      if not(right1<=left2 or left1>=right2 or bottom1>=top2 or top1<=bottom2):
        oc+=1
  return oc

def wire_length(chromosome):
  total_wire=0
  for a,b in connections:
    x1,y1=chromosome[a]
    _,w1,h1=blocks[a]

    centerx1=x1+w1/2
    centery1=y1+h1/2

    x2,y2=chromosome[b]
    _,w2,h2=blocks[b]

    centerx2=x2+w2/2
    centery2=y2+h2/2

    distance=math.sqrt((centerx1-centerx2)**2 + (centery1-centery2)**2)

    total_wire+=distance
  return total_wire

def area (chromosome):
  maxi_x=-math.inf
  mini_x=math.inf
  maxi_y=-math.inf
  mini_y=math.inf

  for (x, y), (_, w, h) in zip(chromosome, blocks):
    mini_x = min(mini_x, x)
    maxi_x = max(maxi_x, x + w)

    mini_y = min(mini_y, y)
    maxi_y = max(maxi_y, y + h)

  area=(maxi_x-mini_x)*(maxi_y-mini_y)
  return area

def select_parent(population):

    candidates = random.sample(population, 3)

    winner = max(candidates, key=fitness)

    return winner

def cross (p1,p2):
  point=random.randint(1,len(p1)-1)
  child1=p1[:point]+p2[point:]
  child2=p2[:point]+p1[point:]

  return child1,child2

def mutate(chromo):
  p=random.randint(0,len(chromo)-1)
  x,y = chromo[p]
  _,w,h=blocks[p]
  x=random.randint(0,25-w)
  y=random.randint(0,25-h)
  chromo[p]=[x,y]
  return chromo

def GA():

  population=[]
  for i in range (6):
    temp=[]
    for n, width, height in blocks:
      x=random.randint(0,25-width)
      y=random.randint(0,25-height)
      temp.append([x,y])
    population.append(temp)
  best=max(population,key=fitness)

  for i in range(15):
    population.sort(key=fitness,reverse=True)
    new_pop=[population[0]]

    while len(new_pop)<6:
      p1=select_parent(population)
      p2=select_parent(population)

      child1,child2=cross(p1,p2)

      if random.randint(1,10)<8:
        child1=mutate(child1)
      if random.randint(1,10)<8:
        child2=mutate(child2)
      new_pop.append(child1)

      if len(new_pop)<Max_pop:
        new_pop.append(child2)

    population=new_pop
    current_b=max(population,key=fitness)

    if fitness(current_b)>fitness(best):
      best=current_b
  print("Best Chromosome:",best)
  print("Best fitness:",fitness(best))
  print("Total wiring length:",wire_length(best))
  print("Bounding box:",area(best))
  print("Overlaps:",overlapping(best))


GA()

````
