# **Lab Evaluation**

```python
import math
tree = {'A': ['B', 'C'],
        'B': ['D', 'E'],
        'C': ['F', 'G'],
        'D': ['H', 'I'],
        'E': ['J', 'K'],
        'F': ['L', 'M'],
        'G': ['N', 'O'] }
pruned=[]

leaf_values = { 'H': -1, 'I': 3, 'J': 5, 'K': 1, 'L': -6, 'M': -4, 'N': 0, 'O': 9 }

def minimax(node, maximizing, alpha, beta):
  if node in leaf_values:
    return leaf_values[node]

  if maximizing:
    best=-math.inf

    for i, child in enumerate (tree[node]):
      value=minimax(child, False, alpha, beta)
      best=max(best, value)
      alpha=max(best, alpha)

      if alpha>=beta:
        pruned.extend(tree[node][i+1:])
        break

    return best

  else:
    best=math.inf

    for i, child in enumerate (tree[node]):
      value=minimax(child, True, alpha, beta)
      best=min(best, value)
      beta=min(beta, best)

      if alpha>=beta:
        pruned.extend(tree[node][i+1:])
        break

    return best

answer=minimax('A', True, -math.inf, math.inf)
print(pruned)
print(answer)
```

# **Lab Assignment-01**

```python
# Input
pool = input().split(',')
target = input().strip()
sid = list(map(int, input().split()))

# Take the last n digits of student ID
n = len(target)
weights = sid[-n:]

pruned = []


# Calculate utility
def utility(gene):

    score = 0

    for i in range(len(gene)):

        # Weight
        if i < len(weights):
            w = weights[i]
        else:
            w = 1

        # Target ASCII
        if i < len(target):
            target_value = ord(target[i])
        else:
            target_value = 0

        gene_value = ord(gene[i])

        score += w * abs(gene_value - target_value)

    return -score


# Minimax + Alpha-Beta
def minimax(pool, gene, maximizing, alpha, beta):

    # No nucleotides left
    if len(pool) == 0:
        return utility(gene), gene

    # MAX player
    if maximizing:

        best_value = -999999
        best_gene = ""

        for i in range(len(pool)):

            nucleotide = pool[i]

            # Remove selected nucleotide
            new_pool = pool[:i] + pool[i+1:]

            value, sequence = minimax(
                new_pool,
                gene + nucleotide,
                False,
                alpha,
                beta
            )

            if value > best_value:
                best_value = value
                best_gene = sequence

            alpha = max(alpha, best_value)

            # Alpha-Beta pruning
            if alpha >= beta:
                break

        return best_value, best_gene

    # MIN player
    else:

        best_value = 999999
        best_gene = ""

        for i in range(len(pool)):

            nucleotide = pool[i]

            # Remove selected nucleotide
            new_pool = pool[:i] + pool[i+1:]

            value, sequence = minimax(
                new_pool,
                gene + nucleotide,
                True,
                alpha,
                beta
            )

            if value < best_value:
                best_value = value
                best_gene = sequence

            beta = min(beta, best_value)

            # Alpha-Beta pruning
            if alpha >= beta:
                break

        return best_value, best_gene


# Start Minimax
value, gene = minimax(
    pool,
    "",
    True,
    -999999,
    999999
)

print("Best gene sequence generated:", gene)
print("Utility score:", value)
```

```python
import math
import random


def strength(x):
    return math.log2(x + 1) + x / 10


def utility(maxV, minV):

    i = random.randint(0, 1)
    r = random.randint(1, 10)

    return (strength(maxV) - strength(minV) + ((-1) ** i) * r / 10)


def minimax(depth, maximizing, alpha, beta, maxV, minV):

    # Leaf node
    if depth == 5:
        return utility(maxV, minV)

    # MAX
    if maximizing:

        best = -float('inf')

        for move in range(2):

            value = minimax(depth + 1,False,alpha,beta,maxV,minV)

            best = max(best, value)

            alpha = max(alpha, best)

            # Alpha-Beta pruning
            if alpha >= beta:
                break

        return best

    # MIN
    else:

        best = float('inf')

        for move in range(2):

            value = minimax(depth + 1,True,alpha,beta,maxV,minV)

            best = min(best, value)

            beta = min(beta, best)

            # Alpha-Beta pruning
            if alpha >= beta:
                break

        return best
```
