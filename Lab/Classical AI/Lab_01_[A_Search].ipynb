```python
#INPUTS
graph = {'A': {'B': 75, 'C': 118, 'E': 140},
         'B': {'A': 75},
         'C': {'A': 118,'D': 111},
         'D': {'C': 111},
         'E': {'A': 140, 'G': 80, 'F': 99},
         'F': {'E': 99, 'I': 211},
         'G': {'E':80, 'H': 97},
         'H': {'G': 97, 'I': 101}}
heuristics = {'A': 366, 'B': 374, 'C': 329, 'D': 244, 'E': 253, 'F': 178, 'G': 193, 'H': 98, 'I': 0}


#Main Function
import heapq
def A_star_search(graph, heuristics, start, goal):
  open = [] #priority queue
  visited = []

  heapq.heappush(open, (0 + heuristics[start], 0, start, [start]))

  while open:
    fn, gn, current, path = heapq.heappop(open)
    if current in visited:
      continue
    visited.append(current)

    if current == goal:
      return path, gn
 # --------Complete this part--------
    for neigh, c in graph[current].items():
      if neigh not in visited:
        new_gn=c+gn
        new_fn=new_gn+heuristics[neigh]
        heapq.heappush(open,(new_fn, new_gn, neigh, path+[neigh]))
  # --------------------------------------------------------

  return None, None

#Driver Code
path, cost = A_star_search(graph, heuristics, 'A', 'I')
print(path, cost) #This will print ['A', 'E', 'G', 'H', 'I'] 418
```

```python
# Calculating heuristic
def heuristic(start, goal):
  return abs(start[0]-goal[0])+abs(start[1]-goal[1])
```

```python
# A* function
import heapq
def Asearch(maze, start, goal, r, c):
  open=[]
  visited=[]
  start_h=heuristic(start, goal)
  heapq.heappush(open,(start_h, 0, start, ""))

  directions=[(-1,0,'U'),(1,0,'D'),(0,1,'R'),(0,-1,'L')]

  while open:
    fn,gn,current,path=heapq.heappop(open)

    if current in visited:
      continue
    visited.append(current)

    if current==goal:
      return gn, path

    row, col=current

    for i, j, dir in directions:
      new_r=row+i
      new_c=col+j


      if(0<=new_r<r and 0<=new_c<c and maze[new_r][new_c]=='0'):

        neighbour=(new_r, new_c)
        if neighbour not in visited:
          new_g=gn+1
          new_f=new_g+heuristic(neighbour,goal)
          heapq.heappush(open,(new_f, new_g, neighbour, path+dir))
  return -1, ""

```

```python
# Input/Output
r,c=map(int,input().split())
st_r, st_c=map(int,input().split())
goal_r, goal_c=map(int,input().split())
maze=[]

for i in range(r):
  row=input()
  maze.append(list(row))

start=(st_r, st_c)
goal=(goal_r, goal_c)
cost, path=Asearch(maze, start, goal, r, c)

if cost==-1:
  print(-1)
else:
  print(cost)
  print(path)
```

```python
from collections import deque
def bfs(graph, start, goal):
  visited=[]
  queue=deque()
  queue.append((start,0))

  while queue:
    current,cost=queue.popleft()

    if current in visited:
      continue
    visited.append(current)

    if current==goal:
      return cost

    for neigh in graph[current]:
      if neigh not in visited:
        queue.append((neigh, cost+1))
  return -1

graph={}
heuristics={}
inadmissible=[]

vertices, edges=map(int, input().split())
start, goal=map(int, input().split())

for i in range (vertices):
  x,h=map(int, input().split())
  heuristics[x]=h
  graph[x]=[]

for i in range (edges):
  x,y=map(int, input().split())
  graph[x].append(y)
  graph[y].append(x)

for node,h in heuristics.items():

  ac=bfs(graph, node, goal)

  if h>ac:
    inadmissible.append(node)

if len(inadmissible)==0:
  print(1)
else:
  print(0)
  print(inadmissible)





```

```python
start=((1, 8, 3),
       (2, 0, 4),
       (7, 6, 5))
goal=((1, 2, 3),
      (4, 5, 6),
      (7, 8, 0))


def heuristic(state,goal):
  distance=0
  for i in range(3):
    for j in range(3):
      if state[i][j]!=0:
        row=(state[i][j]-1)//3
        col=(state[i][j]-1)%3
        distance+=abs(row-i)+abs(col-j)
  return distance


import heapq
def Astar (start, goal):
  visited=set()
  open=[]
  heapq.heappush(open,(heuristic(start, goal), 0, start))
  directions=[(1,0),(-1,0),(0,-1),(0,1)]

  while open:
    fn, gn, current=heapq.heappop(open)

    if current in visited:
      continue
    visited.add(current)

    if current==goal:
      return gn

    for i in range(3):
      for j in range(3):
        if current[i][j]==0:
          row=i
          col=j
          break
    for i,j in directions:
      new_r=row+i
      new_c=col+j

      if 0<=new_r<3 and 0<=new_c<3:

        new_state= [list(x) [:] for x in current]

        new_state[row][col],new_state[new_r][new_c]=new_state[new_r][new_c],new_state[row][col]

        new_tuple = tuple(tuple(x) for x in new_state)

        if new_tuple not in visited:
          new_gn=gn+1
          new_fn=new_gn+heuristic(new_tuple,goal)
          heapq.heappush(open,(new_fn, new_gn, new_tuple))
  return -1

print(Astar(start, goal))

```

```python
graph={'S': [('A', 4), ('B', 2)],
       'B':[('A', 1),('G', 5)],
       'A':[('G', 2)],
       'G':[]}
heuristics={'S':6, 'A': 2, 'B':4, 'G':0}
import heapq

def Astar(start, goal):
  visited=[]
  open=[]
  heapq.heappush(open,(heuristics[start]+0, 0, start, [start]))

  while open:
    fn, gn, current, path=heapq.heappop(open)

    if current in visited:
      continue
    visited.append(current)

    if current==goal:
      return path

    for neigh, cost in graph[current]:
      if neigh not in visited:
        new_gn=gn+cost
        new_fn=new_gn+heuristics[neigh]
        heapq.heappush(open,(new_fn, new_gn, neigh, path+[neigh]))
  return -1

Astar('S', 'G')
```
